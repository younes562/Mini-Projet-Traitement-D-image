# SURF (Speeded-Up Robust Features)

## Objectif
Étudier l’algorithme SURF pour l’extraction de caractéristiques.

Pipeline théorique :
Images → SURF → Descripteurs → Classification

## Principe de SURF

SURF est un algorithme basé sur la détection de points clés (keypoints)
et l’extraction de descripteurs robustes.

Il est plus rapide que SIFT et robuste aux transformations.

## Problème rencontré



L’algorithme SURF n’a pas pu être implémenté en pratique.

Erreur rencontrée :
AttributeError: module 'cv2' has no attribute 'xfeatures2d'

Raison :
SURF est un algorithme breveté et n’est plus inclus dans certaines versions d’OpenCV.

##  Solution

Pour remplacer SURF, nous avons utilisé ORB (Oriented FAST and Rotated BRIEF),
qui est une alternative libre et rapide.
