# Comparaison des méthodes : HOG vs LBP vs SIFT

## Objectif
Comparer les performances des différentes méthodes d’extraction de caractéristiques
utilisées pour la classification des images (cats vs dogs).

##  Résultats

```python
import pandas as pd

data = {
    "Méthode": ["HOG", "LBP", "SIFT"],
    "Accuracy": [0.72, 0.60, 0.60],
    "Precision (avg)": [0.73, 0.60, 0.60],
    "Recall (avg)": [0.72, 0.60, 0.60],
    "F1-score (avg)": [0.72, 0.60, 0.60]
}

df = pd.DataFrame(data)
df
```

## Graph

```python
import matplotlib.pyplot as plt

plt.bar(df["Méthode"], df["Accuracy"])
plt.title("Comparaison des Accuracy")
plt.xlabel("Méthode")
plt.ylabel("Accuracy")
plt.show()
```

## Analyse des résultats

- HOG donne la meilleure performance avec une accuracy de 72%.
- LBP et SIFT donnent des résultats similaires (~60%).

### Interprétation

- HOG est efficace pour capturer les formes et contours des objets (chiens et chats).
- LBP est basé sur la texture, ce qui est moins discriminant pour ce type d’images.
- SIFT, رغم أنه قوي، تم تبسيطه (mean descriptors)، مما أثر على الأداء.

## Conclusion finale

Dans ce projet, nous avons comparé trois méthodes :

- HOG : meilleure performance
- LBP : rapide mais moins précis
- SIFT : robuste mais performance moyenne ici

### Conclusion générale

HOG est la méthode la plus adaptée pour ce problème de classification (cats vs dogs).
